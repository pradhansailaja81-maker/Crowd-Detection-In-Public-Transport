# Check supported image formats

# Identify corrupted images

```python
import cv2

corrupted_images = []

for split in ["train", "val", "test"]:
    folder = IMAGE_PATH / split

    if folder.is_dir():
        for file in folder.iterdir():

            if not file.is_file() or file.suffix.lower() not in supported_formats:
                continue

            image = cv2.imread(str(file))

            if image is None:
                corrupted_images.append({
                    "Split": split,
                    "Image": file.name,
                    "Reason": "Cannot read or decode image"
                })

corrupted_df = pd.DataFrame(
    corrupted_images,
    columns=["Split", "Image", "Reason"]
)

print("Corrupted images:", len(corrupted_df))
display(corrupted_df.head(20))
```

# Identify duplicate images

```python
import hashlib

image_hashes = {}
duplicate_images = []

for split in ["train", "val", "test"]:
    folder = IMAGE_PATH / split

    if folder.is_dir():
        for file in folder.iterdir():

            if not file.is_file() or file.suffix.lower() not in supported_formats:
                continue

            try:
                with file.open("rb") as image_file:
                    file_hash = hashlib.sha256(image_file.read()).hexdigest()

                if file_hash in image_hashes:
                    duplicate_images.append({
                        "Split": split,
                        "Duplicate Image": file.name,
                        "Original Image": image_hashes[file_hash]
                    })
                else:
                    image_hashes[file_hash] = f"{split}/{file.name}"

            except OSError as error:
                print("Could not inspect:", file.name, error)

duplicate_df = pd.DataFrame(
    duplicate_images,
    columns=["Split", "Duplicate Image", "Original Image"]
)

print("Duplicate images found:", len(duplicate_df))
display(duplicate_df.head(20))
```

# Check missing label files

```python
LABEL_PATH = ORIGINAL_DATASET_PATH / "labels"

missing_labels = []
orphan_labels = []

for split in ["train", "val", "test"]:

    image_folder = IMAGE_PATH / split
    label_folder = LABEL_PATH / split

    if image_folder.is_dir():

        image_files = [
            f for f in image_folder.iterdir()
            if f.is_file() and f.suffix.lower() in supported_formats
        ]

        label_stems = (
            {f.stem for f in label_folder.glob("*.txt")}
            if label_folder.is_dir()
            else set()
        )

        image_stems = {f.stem for f in image_files}

        for image_file in image_files:
            if image_file.stem not in label_stems:
                missing_labels.append({
                    "Split": split,
                    "Image": image_file.name,
                    "Expected Label": image_file.stem + ".txt"
                })

        for label_file in (
            label_folder.glob("*.txt") if label_folder.is_dir() else []
        ):
            if label_file.stem not in image_stems:
                orphan_labels.append({
                    "Split": split,
                    "Label": label_file.name
                })

missing_df = pd.DataFrame(
    missing_labels,
    columns=["Split", "Image", "Expected Label"]
)

orphan_df = pd.DataFrame(
    orphan_labels,
    columns=["Split", "Label"]
)

print("Images with missing labels:", len(missing_df))
display(missing_df.head(20))

print("Labels without matching images:", len(orphan_df))
display(orphan_df.head(20))
```

# Validate bounding boxes

```python
invalid_boxes = []
valid_box_count = 0
total_box_count = 0
empty_label_files = 0

# Get the number of classes from data.yaml if available
class_count = None

if yaml_path.is_file():
    with yaml_path.open("r", encoding="utf-8") as file:
        config = yaml.safe_load(file)

    class_count = config.get("nc")

for split in ["train", "val", "test"]:

    label_folder = LABEL_PATH / split

    if not label_folder.is_dir():
        continue

    for label_file in label_folder.glob("*.txt"):

        try:
            lines = label_file.read_text(encoding="utf-8").splitlines()

            if not any(line.strip() for line in lines):
                empty_label_files += 1
                continue

            for line_number, line in enumerate(lines, start=1):

                if not line.strip():
                    continue

                total_box_count += 1
                values = line.split()

                reason = None

                if len(values) != 5:
                    reason = "Incorrect number of values"

                else:
                    try:
                        class_value = float(values[0])
                        x, y, w, h = map(float, values[1:])

                        if not all(
                            np.isfinite(v)
                            for v in [class_value, x, y, w, h]
                        ):
                            reason = "Non-finite coordinate"

                        elif (
                            class_value < 0
                            or not class_value.is_integer()
                        ):
                            reason = "Invalid class ID"

                        elif class_count is not None and class_value >= class_count:
                            reason = "Class ID outside configured range"

                        elif not (0 <= x <= 1 and 0 <= y <= 1):
                            reason = "Center outside normalized range"

                        elif not (0 < w <= 1 and 0 < h <= 1):
                            reason = "Invalid width or height"

                        elif (
                            x - w / 2 < -1e-6
                            or y - h / 2 < -1e-6
                            or x + w / 2 > 1 + 1e-6
                            or y + h / 2 > 1 + 1e-6
                        ):
                            reason = "Bounding box extends outside image"

                    except ValueError:
                        reason = "Non-numeric annotation value"

                if reason:
                    invalid_boxes.append({
                        "Split": split,
                        "Label File": label_file.name,
                        "Line": line_number,
                        "Reason": reason,
                        "Content": line
                    })
                else:
                    valid_box_count += 1

        except (OSError, UnicodeError) as error:
            invalid_boxes.append({
                "Split": split,
                "Label File": label_file.name,
                "Line": None,
                "Reason": f"Cannot read label: {error}",
                "Content": ""
            })

invalid_box_df = pd.DataFrame(
    invalid_boxes,
    columns=["Split", "Label File", "Line", "Reason", "Content"]
)

print("Total annotation rows:", total_box_count)
print("Valid bounding boxes:", valid_box_count)
print("Invalid annotation rows:", len(invalid_box_df))
print("Empty label files:", empty_label_files)

display(invalid_box_df.head(20))
```

```python
import yaml
import pandas as pd

yaml_path = ORIGINAL_DATASET_PATH / "data.yaml"

with yaml_path.open("r", encoding="utf-8") as file:
    data_config = yaml.safe_load(file)

class_names = data_config.get("names", [])
number_of_classes = data_config.get("nc")

print("Number of classes:", number_of_classes)
print("Class names:", class_names)
```

```python
class_id_counts = {}

for split in ["train", "val", "test"]:
    folder = LABEL_PATH / split

    if not folder.is_dir():
        continue

    for label_file in folder.glob("*.txt"):
        try:
            for line in label_file.read_text(encoding="utf-8").splitlines():
                values = line.split()

                if len(values) == 5:
                    try:
                        class_value = float(values[0])

                        if class_value.is_integer() and class_value >= 0:
                            class_id = int(class_value)
                            class_id_counts[class_id] = (
                                class_id_counts.get(class_id, 0) + 1
                            )
                    except ValueError:
                        pass

        except (OSError, UnicodeError):
            continue

class_df = pd.DataFrame(
    [
        {
            "Class ID": class_id,
            "Class Name": (
                class_names.get(class_id, class_names.get(str(class_id), "Unknown"))
                if isinstance(class_names, dict)
                else class_names[class_id]
                if isinstance(class_names, list) and class_id < len(class_names)
                else "Unknown"
            ),
            "Annotation Count": count,
            "Valid ID": class_id < number_of_classes
        }
        for class_id, count in sorted(class_id_counts.items())
    ]
)

display(class_df)
```

```python
# Read class names from the existing dataset configuration
class_names_config = data_config.get("names", {})

def get_class_name(class_id):

    if isinstance(class_names_config, dict):
        return str(
            class_names_config.get(
                class_id,
                class_names_config.get(str(class_id), f"Class {class_id}")
            )
        )

    if isinstance(class_names_config, list):
        if 0 <= class_id < len(class_names_config):
            return str(class_names_config[class_id])

    return f"Class {class_id}"


class_counts = {}
total_annotations = 0
invalid_class_ids = 0

for split in ["train", "val", "test"]:

    label_folder = PREPARED_DATASET_PATH / "labels" / split

    if not label_folder.is_dir():
        continue

    for label_file in label_folder.glob("*.txt"):

        with label_file.open("r", encoding="utf-8") as file:
            for line in file:

                values = line.strip().split()

                if len(values) != 5:
                    continue

                try:
                    class_id = int(float(values[0]))
                except ValueError:
                    continue

                if class_id < 0 or class_id >= number_of_classes:
                    invalid_class_ids += 1
                    continue

                class_counts[class_id] = class_counts.get(class_id, 0) + 1
                total_annotations += 1


class_df = pd.DataFrame([
    {
        "Class ID": class_id,
        "Class Name": get_class_name(class_id),
        "Annotations": count
    }
    for class_id, count in sorted(class_counts.items())
])

display(class_df)

print("Total valid class annotations:", total_annotations)
print("Invalid class ID entries:", invalid_class_ids)
```

# 5.4 Analyze bounding-box sizes

```python
bbox_results = []
invalid_bbox_entries = 0

for split in ["train", "val", "test"]:

    image_folder = PREPARED_DATASET_PATH / "images" / split
    label_folder = PREPARED_DATASET_PATH / "labels" / split

    if not image_folder.is_dir() or not label_folder.is_dir():
        continue

    for label_file in label_folder.glob("*.txt"):

        image_path = next(
            (
                image_folder / (label_file.stem + ext)
                for ext in IMAGE_EXTENSIONS
                if (image_folder / (label_file.stem + ext)).is_file()
            ),
            None
        )

        if image_path is None:
            continue

        try:
            with Image.open(image_path) as img:
                image_width, image_height = img.size
        except Exception:
            continue

        with label_file.open("r", encoding="utf-8") as file:

            for line in file:

                values = line.strip().split()

                if len(values) != 5:
                    invalid_bbox_entries += 1
                    continue

                try:
                    class_id = int(float(values[0]))
                    x_center, y_center, box_width, box_height = map(
                        float, values[1:]
                    )
                except ValueError:
                    invalid_bbox_entries += 1
                    continue

                if (
                    class_id < 0
                    or class_id >= number_of_classes
                    or not all(np.isfinite(v) for v in
                               [x_center, y_center, box_width, box_height])
                    or not (0 <= x_center <= 1)
                    or not (0 <= y_center <= 1)
                    or not (0 < box_width <= 1)
                    or not (0 < box_height <= 1)
                    or x_center - box_width / 2 < 0
                    or x_center + box_width / 2 > 1
                    or y_center - box_height / 2 < 0
                    or y_center + box_height / 2 > 1
                ):
                    invalid_bbox_entries += 1
                    continue

                bbox_results.append({
                    "Split": split,
                    "Class": get_class_name(class_id),
                    "Width Normalized": box_width,
                    "Height Normalized": box_height,
                    "Width Pixels": box_width * image_width,
                    "Height Pixels": box_height * image_height,
                    "Area Pixels": (
                        box_width * image_width *
                        box_height * image_height
                    )
                })


bbox_df = pd.DataFrame(bbox_results)

display(bbox_df.head(10))

print("Valid bounding boxes:", len(bbox_df))
print("Invalid bounding-box entries:", invalid_bbox_entries)

if not bbox_df.empty:
    display(bbox_df[
        ["Width Pixels", "Height Pixels", "Area Pixels"]
    ].describe().round(2))
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

INVALID_REPORT = []

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp"}

for split in ["train", "val", "test"]:

    image_folder = PREPARED_DATASET_PATH / "images" / split
    label_folder = PREPARED_DATASET_PATH / "labels" / split

    if not label_folder.is_dir():
        continue

    for label_file in label_folder.glob("*.txt"):

        image_path = next(
            (
                image_folder / (label_file.stem + ext)
                for ext in IMAGE_EXTENSIONS
                if (image_folder / (label_file.stem + ext)).is_file()
            ),
            None
        )

        if image_path is None:
            continue

        try:
            with Image.open(image_path) as img:
                image_width, image_height = img.size
        except Exception:
            continue

        with label_file.open("r", encoding="utf-8") as file:

            for line_number, line in enumerate(file, start=1):

                values = line.strip().split()

                reason = None

                if len(values) != 5:
                    reason = "Incorrect number of values"

                else:
                    try:
                        class_id = int(float(values[0]))
                        x, y, w, h = map(float, values[1:])

                        if not all(np.isfinite(v) for v in [x, y, w, h]):
                            reason = "Non-finite coordinate"

                        elif class_id < 0 or class_id >= number_of_classes:
                            reason = "Invalid class ID"

                        elif not (0 <= x <= 1 and 0 <= y <= 1):
                            reason = "Center outside image"

                        elif not (0 < w <= 1 and 0 < h <= 1):
                            reason = "Invalid width or height"

                        elif (
                            x - w / 2 < 0 or x + w / 2 > 1
                            or y - h / 2 < 0 or y + h / 2 > 1
                        ):
                            reason = "Bounding box extends outside image"

                    except (ValueError, OverflowError):
                        reason = "Non-numeric annotation"

                if reason:
                    INVALID_REPORT.append({
                        "Split": split,
                        "Image": image_path.name,
                        "Label File": label_file.name,
                        "Line Number": line_number,
                        "Annotation": line.strip(),
                        "Reason": reason
                    })

# Create report
invalid_df = pd.DataFrame(INVALID_REPORT)

report_path = REPORTS_PATH / "phase6_invalid_annotations.csv"

invalid_df.to_csv(report_path, index=False)

print("=" * 50)
print("INVALID ANNOTATION INSPECTION")
print("=" * 50)

print("Total invalid entries found:", len(invalid_df))
print("Affected annotation files:", invalid_df["Label File"].nunique() if not invalid_df.empty else 0)

if not invalid_df.empty:
    display(invalid_df.head(20))
    print("\nInvalid annotation reasons:")
    display(invalid_df["Reason"].value_counts().to_frame("Count"))

print("\nReport saved:", report_path)
print("Original dataset unchanged.")
print("=" * 50)
```

# Measure the boundary overflow

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Read the inspection report
report_path = PROJECT_PATH / "outputs" / "reports" / "phase6_invalid_annotations.csv"
check_df = pd.read_csv(report_path)

overflow_values = []

for _, row in check_df.iterrows():
    values = [float(x) for x in row["Annotation"].split()]

    class_id, xc, yc, width, height = values

    # Calculate bounding-box boundaries
    x1 = xc - width / 2
    y1 = yc - height / 2
    x2 = xc + width / 2
    y2 = yc + height / 2

    # Calculate how far the box exceeds the image
    overflow = max(
        0,
        -x1,
        -y1,
        x2 - 1,
        y2 - 1
    )

    overflow_values.append(overflow)

check_df["Overflow"] = overflow_values

print("BOUNDARY OVERFLOW ANALYSIS")
print("=" * 50)

print("Total invalid boxes:", len(check_df))
print("Maximum overflow:", check_df["Overflow"].max())

print("\nOverflow categories:")
print("Tiny overflow (<= 0.00001):",
      (check_df["Overflow"] <= 0.00001).sum())

print("Small overflow (0.00001 to 0.001):",
      ((check_df["Overflow"] > 0.00001) &
       (check_df["Overflow"] <= 0.001)).sum())

print("Larger overflow (> 0.001):",
      (check_df["Overflow"] > 0.001).sum())

print("\nLargest 15 overflow values:")
display(
    check_df.sort_values("Overflow", ascending=False)[
        ["Split", "Image", "Line Number", "Annotation", "Overflow"]
    ].head(15)
)
```

# Validate the cleaned dataset

```python
from pathlib import Path
import yaml
import numpy as np

print("CLEANED DATASET VALIDATION")
print("=" * 55)

# Load cleaned YAML
with open(CLEANED_YAML_PATH, "r") as file:
    clean_config = yaml.safe_load(file)

dataset_root = Path(clean_config.get("path", CLEANED_DATASET_PATH))

if not dataset_root.is_absolute():
    dataset_root = CLEANED_YAML_PATH.parent / dataset_root

print("Cleaned dataset:", CLEANED_DATASET_PATH)
print("YAML file available:", CLEANED_YAML_PATH.is_file())
print("Dataset root available:", dataset_root.is_dir())

image_extensions = {".jpg", ".jpeg", ".png", ".bmp"}

total_images = 0
total_labels = 0
total_boxes = 0
invalid_boxes = 0
missing_labels = 0
orphan_labels = 0

for split in ["train", "val", "test"]:

    image_folder = dataset_root / clean_config[split]
    label_folder = dataset_root / "labels" / split

    if not image_folder.is_absolute():
        image_folder = dataset_root / image_folder

    images = [
        p for p in image_folder.iterdir()
        if p.is_file() and p.suffix.lower() in image_extensions
    ] if image_folder.is_dir() else []

    labels = list(label_folder.glob("*.txt")) if label_folder.is_dir() else []

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in labels}

    missing = image_stems - label_stems
    orphan = label_stems - image_stems

    missing_labels += len(missing)
    orphan_labels += len(orphan)

    split_boxes = 0
    split_invalid = 0

    for label_file in labels:
        for line in label_file.read_text().splitlines():

            if not line.strip():
                continue

            parts = line.split()

            try:
                if len(parts) != 5:
                    raise ValueError

                class_id = int(parts[0])
                xc, yc, w, h = map(float, parts[1:])

                if (
                    not np.isfinite([xc, yc, w, h]).all()
                    or class_id < 0
                    or class_id >= clean_config["nc"]
                    or w <= 0 or h <= 0
                    or xc < 0 or xc > 1
                    or yc < 0 or yc > 1
                    or xc - w / 2 < -1e-9
                    or yc - h / 2 < -1e-9
                    or xc + w / 2 > 1 + 1e-9
                    or yc + h / 2 > 1 + 1e-9
                ):
                    raise ValueError

                split_boxes += 1

            except (ValueError, TypeError):
                split_invalid += 1

    total_images += len(images)
    total_labels += len(labels)
    total_boxes += split_boxes
    invalid_boxes += split_invalid

    print(f"\n{split.upper()} SPLIT")
    print("Images:", len(images))
    print("Labels:", len(labels))
    print("Missing labels:", len(missing))
    print("Orphan labels:", len(orphan))
    print("Valid boxes:", split_boxes)
    print("Invalid boxes:", split_invalid)

print("\nFINAL VALIDATION SUMMARY")
print("=" * 55)
print("Total images:", total_images)
print("Total label files:", total_labels)
print("Total valid boxes:", total_boxes)
print("Invalid boxes:", invalid_boxes)
print("Missing labels:", missing_labels)
print("Orphan labels:", orphan_labels)

validation_passed = (
    total_images == 2122
    and invalid_boxes == 0
    and missing_labels == 0
    and orphan_labels == 0
    and dataset_root.is_dir()
)

print("\nDATASET VALIDATION:",
      "PASSED" if validation_passed else "NEEDS REVIEW")
```

# Phase 9 – Crowd Level Analysis

```python
# STEP 9.1: DEFINE CROWD-LEVEL THRESHOLDS

LOW_MAX = 5
MEDIUM_MAX = 15

print("CROWD LEVEL THRESHOLDS")
print("=" * 40)
print("Low Crowd:     0 to", LOW_MAX, "people")
print("Medium Crowd:  6 to", MEDIUM_MAX, "people")
print("High Crowd:    16 or more people")
```

# Calculate crowd count

```python
# STEP 9.2: CALCULATE CROWD COUNT

import pandas as pd

if "person_count_df" not in globals():
    raise NameError(
        "person_count_df not found. Complete Step 8.4 first."
    )

if not {"Image", "Detected_Persons"}.issubset(person_count_df.columns):
    raise ValueError("Required person-count columns are missing.")

crowd_analysis_df = person_count_df.copy()

crowd_analysis_df["Detected_Persons"] = pd.to_numeric(
    crowd_analysis_df["Detected_Persons"],
    errors="raise"
)

if (crowd_analysis_df["Detected_Persons"] < 0).any():
    raise ValueError("Person counts cannot be negative.")

print("STEP 9.2: CROWD COUNT CALCULATED")
display(crowd_analysis_df)
```

# Classify Low, Medium and High crowd levels

```python
# STEP 9.3: CROWD LEVEL CLASSIFICATION

def classify_crowd(count):

    if count <= LOW_MAX:
        return "Low"

    elif count <= MEDIUM_MAX:
        return "Medium"

    else:
        return "High"


crowd_analysis_df["Crowd_Level"] = (
    crowd_analysis_df["Detected_Persons"].apply(classify_crowd)
)

print("STEP 9.3: CROWD CLASSIFICATION")
print("=" * 45)

display(crowd_analysis_df)
```

# Analyze crowd levels image by image

```python
# STEP 9.4: IMAGE-WISE CROWD ANALYSIS

print("STEP 9.4: IMAGE-WISE CROWD ANALYSIS")
print("=" * 50)

for index, row in crowd_analysis_df.iterrows():

    print("Image:", row["Image"])
    print("Person count:", row["Detected_Persons"])
    print("Crowd level:", row["Crowd_Level"])
    print("-" * 35)
```

# Generate crowd statistics

```python
# STEP 9.5: GENERATE CROWD STATISTICS

if crowd_analysis_df.empty:
    raise ValueError("No image records available for analysis.")

total_images = len(crowd_analysis_df)

average_count = crowd_analysis_df["Detected_Persons"].mean()

maximum_count = crowd_analysis_df["Detected_Persons"].max()

minimum_count = crowd_analysis_df["Detected_Persons"].min()

level_distribution = (
    crowd_analysis_df["Crowd_Level"]
    .value_counts()
    .reindex(["Low", "Medium", "High"], fill_value=0)
)

crowd_statistics = pd.DataFrame({
    "Statistic": [
        "Total Images Analyzed",
        "Average Detected Persons",
        "Maximum Detected Persons",
        "Minimum Detected Persons",
        "Low Crowd Images",
        "Medium Crowd Images",
        "High Crowd Images"
    ],
    "Value": [
        total_images,
        round(average_count, 2),
        int(maximum_count),
        int(minimum_count),
        int(level_distribution["Low"]),
        int(level_distribution["Medium"]),
        int(level_distribution["High"])
    ]
})

print("STEP 9.5: CROWD STATISTICS")
print("=" * 45)

display(crowd_statistics)
```
